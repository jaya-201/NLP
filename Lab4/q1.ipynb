```python
import pandas as pd
from collections import Counter, defaultdict
```

```python
df = pd.read_parquet("../Assignment_1/tokenized_hi.parquet")
```

```python
all_tokens = df['sentences'].explode().str['tokens'].explode().tolist()
print("Total individual tokens:", len(all_tokens))
print("First 30 individual tokens:", all_tokens[:30])
```

```python
def build_ngram_model(tokens, n=1):
    """Build n-gram language model for Hindi tokens"""
    ngrams = []
    for i in range(len(tokens) - n + 1):
        ngram = tuple(tokens[i:i+n])
        ngrams.append(ngram)

    counts = Counter(ngrams) #counts frequency of each ngram.
    model = defaultdict(dict)

    for ngram, freq in counts.items(): #maps prefix -> {next_word: frequency}
        prefix = ngram[:-1] if n > 1 else ()
        model[prefix][ngram[-1]] = freq

    # Normalize to probabilities (conditional probablity model)
    for prefix in model:
        total_prefix = sum(model[prefix].values())
        for word in model[prefix]:
            model[prefix][word] /= total_prefix

    return model
```

```python
unigram_model = build_ngram_model(all_tokens, 2)
print("\nUnigram probabilities (top 10):")
print(dict(list(unigram_model[()].items())[:10]))
```

```python
bigram_model = build_ngram_model(all_tokens, 3)
print("\nBigram probabilities after ('लोगों',):")
print(bigram_model[("लोगों",)])
```

```python
trigram_model = build_ngram_model(all_tokens, 4)
print("\nTrigram probabilities after ('भारत','की'):")
print(trigram_model[("भारत","की")])
```

```python
print("\nTrigram probabilities after ('जहां','आई'):")
print(trigram_model[("जहां","आई")])
```

```python
quadgram_model = build_ngram_model(all_tokens, 5)
print("\nQuadrigram probabilities after ('भारत','की','राजधानी'):")
print(quadgram_model[("भारत","की","राजधानी")])
```
